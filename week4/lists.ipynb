```python
english_mark_1=67
english_mark_2=78
english_mark_3=88
english_mark_4=90
english_mark_5=95
english_mark_6=100

total=english_mark_1 + english_mark_2 + english_mark_3 + english_mark_4 + english_mark_5 + english_mark_6
average_mark=total/6

print(average_mark)

if average_mark >= 90:
    print("Grade: A")
elif average_mark >= 80:
    print("Grade: B")
elif average_mark >= 70:
    print("Grade: C")
elif average_mark >= 60:
    print("Grade: D")
elif average_mark >= 50:
    print("Grade: E")
else:
    print("Grade: F")
    
```
