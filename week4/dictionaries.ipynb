```python
student={
    'name':'John',
    'age':20,
    'courses':['Math','CompSci'],
    'fav_num': {1, 2, 3}
}

print(student['name'])
print(student['age'])
print(student['courses'])
print(student['fav_num'])

#getting the last course student took

print(student['courses'][-1])
```

```python
student={
    'first_name':'Jane',
    'last_name':'Doe',
    'married':False,
    'age':20,
    'courses':['Math','CompSci'],
    'fav_num': {1, 2, 3}
}

student['married']= True
student['last_name']='Smith'

print(student)
print(student['married'])
print(student['last_name'])
```
