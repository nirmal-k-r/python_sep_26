```python
#library system
#library class
#member class
#staff class
#shelf class
#book class

class Library:
    def __init__(self, name, address):
        self.name = name
        self.address = address
        self.shelves = []
        self.staff = []
        self.members = []

    def add_shelf(self, shelf):
        self.shelves.append(shelf)

    def add_staff(self, staff_member):
        self.staff.append(staff_member)

    def add_member(self, member):
        self.members.append(member)

    def search_books(self,title):
            for shelf in self.shelves:
                book = shelf.search_book(title)
                if book is not None:
                    return book,shelf
            return None,None
    

    def info(self):
        print(f"Library Name: {self.name}, Address: {self.address}")
        print(f"Shelves: {len(self.shelves)}, Staff: {len(self.staff)}, Members: {len(self.members)}")  

   
class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn

    def info(self):
        print(f"Book Title: {self.title}, Author: {self.author}, ISBN: {self.isbn}")

class Member:
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id

    def search_book(self,library,title,id="none"):
        book,shelf = library.search_books(title)
        if book is not None:
            print(f"Book '{book.title}' found in shelf {shelf.number}.")
            return book,shelf
        else:
            print(f"Book '{title}' not found in the library.")
        return None,None

    def rent_book(self,book,shelf,library):
        book,shelf = self.search_book(library, book.title)
        shelf.remove_book(book)
        print(f"Book '{book.title}' rented out from shelf {shelf.number} by {self.name}.")

    def return_book(self,book,shelf):
        shelf.add_book(book)
        print(f"Book '{book.title}' returned to shelf {shelf.number} by {self.name}.")

    def info(self):
        print(f"Member Name: {self.name}, Member ID: {self.member_id}")

class Staff:
    def __init__(self, name,salary):
        self.name = name
        self.salary = salary
        self.records=[]

    def add_record(self, record):
        self.records.append(record)

    def __str__(self):
        return f"Staff Name: {self.name}, Salary: {self.salary}"



class Shelf:
    def __init__(self, number, category):
        self.number = number
        self.category = category
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def remove_book(self, book):
        self.books.remove(book)

    def search_book(self,title):
        for book in self.books:
            if book.title.lower() in title.lower():
                return book
        return None
```

```python
#initialise the library
lib=Library("Central Library", "123 Main St")

#add shelves to the library
shelf1 = Shelf(1, "Fiction")
shelf2 = Shelf(2, "Non-Fiction")
lib.add_shelf(shelf1)
lib.add_shelf(shelf2)

#add staff
paul=Staff("Paul", 50000)
lib.add_staff(paul)

#add members
alice=Member("Alice", 101)
bob=Member("Bob", 102)
lib.add_member(alice)
lib.add_member(bob)


#add books to shelves
book1 = Book("The Great Gatsby", "F. Scott Fitzgerald", "9780743273565")
book2 = Book("To Kill a Mockingbird", "Harper Lee", "9780061120084")
book3 = Book("Sapiens: A Brief History of Humankind", "Yuval Noah Harari", "9780062316097")
book4 = Book("Educated", "Tara Westover", "9780399590504")
book5 = Book("1984", "George Orwell", "9780451524935")
book6 = Book("The Catcher in the Rye", "J.D. Salinger", "9780316769488")


#add books to shelves
shelf1.add_book(book1)
shelf1.add_book(book2)
shelf2.add_book(book3)
shelf2.add_book(book4)
shelf1.add_book(book5)
shelf1.add_book(book6)


#search for a book
book1,shelf = alice.search_book(lib, "1984")

if book1 is not None:
    #rent the book
    alice.rent_book(book1,shelf,lib)

book,shelf = bob.search_book(lib, "1984")

#return the book
alice.return_book(book1,shelf1)

book,shelf = bob.search_book(lib, "1984")
bob.rent_book(book,shelf,lib)
```

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def info(self):
        print(f"Car Make: {self.make}, Model: {self.model}, Year: {self.year}")


class ElectricCar(Car): #inherit from Car class
    pass


myCar=ElectricCar("Tesla", "Model S", 2022)
myCar.info()  # This will call the info method from the Car class
```

```python
class Car: #base class/parent class/ super class
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def info(self): 
        print(f"Car Make: {self.make}, Model: {self.model}, Year: {self.year}")


class ElectricCar(Car): #inherit from Car class - child class/subclass
    def __init__(self, make, model, year, battery_capacity):
        Car.__init__(self,make, model, year)  # Call the constructor of the parent class
        self.battery_capacity = battery_capacity

    def info(self): #polymorphism - overriding the info method of the parent class
        Car.info(self)  # Call the info method from the parent class
        print(f"Battery Capacity: {self.battery_capacity} kWh")  # Print the battery capacity


myCar=ElectricCar("Tesla", "Model S", 2022,98)
myCar.info()  # This will call the info method from the Car class
```

```python
class Rectangle:
    def __init__(self,length,width):
        self.length=length
        self.width=width

    @property
    def area(self):
        return self.length * self.width

    #setter for area property
    @area.setter
    def area(self,value):
        self.length=value/self.width

    def __str__(self):
        return f"Rectangle(length={self.length}, width={self.width})"


x=Rectangle(10,5)
print(x.area)  # Accessing the area property


x.area=100
print(x)  # This will call the __str__ method and print the updated length and width
```

```python
class Calculator:
    @staticmethod
    def add(x, y):
        return x + y

result=Calculator.add(5, 3)  # Calling the static method without creating an instance
print(result)  # Output: 8
```
