### **Exercise 1: Simple If Statement**

- **Objective**: Write a program that checks if a number is positive. If it is, print "The number is positive."
- **Hint**: Use a simple `if` statement.

```python
num=10


if num>0:
    print("Number is positive")
else:
    print("Number is negative")
```

```python
married=False


if (married==False):
    print("You are not married")
else:
    print("You are married")


# short hand if else statement
print("You are not married") if married == False else print("You are married")
```
