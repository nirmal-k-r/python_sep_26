```python
#Buyuilding a clculator program

def add(num1,num2):
    total=num1+num2
    print(total)

def subtract(num1,num2):
    total=num1-num2
    print(total)

def multiply(num1,num2):
    total=num1*num2
    print(total)

def divide(num1,num2):
    try:
        total=num1/num2
        print(total)
    except:
        print("Error: Division by zero is not allowed")
    

def calculator():
    num1=int(input("Enter first number: "))
    num2=int(input("Enter second number: "))
    operation=input("Enter operation (+,-,*,/): ")
    
    if operation=='+':
        add(num1,num2)
    elif operation=='-':
        subtract(num1,num2)
    elif operation=='*':
        multiply(num1,num2) 
    elif operation=='/':
        divide(num1,num2)
    else:
        print("Invalid operation")


#main program
calculator()
```

```python
#variable length arguments - *args 


def avg(*nums):
    # if (len(nums)==0):
    #     print("No numbers provided")
    #     return
    total=0
    for num in nums:
        total+=num
    myAvg=total/len(nums)
    print(myAvg)


# avg()
avg(10)
avg(10,20)
avg(10,20,30)
avg(10,20,30,40)
avg(10,20,30,40,50)
```

```python
#keyword arguments - **kwargs

def citizen_info(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

citizen_info(name="John Doe")
citizen_info(name="John Doe", age=30)
citizen_info(name="John Doe", age=30, gender="Male", address="123 Main St")
```
