### exceptions
_________
- an error that causes the program to stop running. Exceptions are raised when an error occurs during the execution of a program.


### errors
__________
- syntax errors: an error in the grammar of the language - pront insted of print
- logical/semantic errors: errors that are logic bounded - division by zero, read a file that does not exist

```python
#valid and correct code
print('Program starts')
a=10
b=1
c=a/b
print(c)
print('Program ends')
```

```python
#introduce a logical error
print('Program starts')
a=10
b=0
c=a/b
print(c)
print('Program ends')
```

```python
#wrap error prone code in try except block
print('Program starts')
a=10
b=0

try:
    c=a/b
    print(c)
except: #only executes when error occurs
    print("Error occurred while dividing")  


print('Program ends')
```

```python
#no error occurs
print('Program starts')
a=10
b=1

try:
    c=a/b
    print(c)
except: #only executes when error occurs
    print("Error occurred while dividing")  

print('Program ends')
```

```python
#no error occurs
print('Program starts')
a=10
b=0

try:
    c=a/b
    print(c)
except Exception as e: #only executes when error occurs
    print(str(e))  

print('Program ends')
```

```python
#no error occurs
print('Program starts')
a=10
b=1

try:
    c=a/b
except: #only executes when error occurs
    print('error occurred')  
else: #only executes when no error occurs
    print("No error occurred")
    print(c)

print('Program ends')
```

```python
#no error occurs
print('Program starts')
a=10
b=0

try:
    c=a/b
except: #only executes when error occurs
    print('error occurred')  
else: #only executes when no error occurs
    print("No error occurred")
    print(c)
finally: #executes always regardless of error or not
    print('Program ends')
```
